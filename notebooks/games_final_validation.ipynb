# `games_final.csv` testing and validation

```python
import pandas as pd
matched = pd.read_csv('../data/processed/games_final.csv')
matched['GAME_DATE'] = pd.to_datetime(matched['GAME_DATE'])
```

```python
print("Duplicates GAME_ID:", matched['GAME_ID'].duplicated().sum())
print("Date range:", matched['GAME_DATE'].min(), "-", matched['GAME_DATE'].max())
print("Number of unique seasons:", matched['SEASON_ID'].nunique())
print(matched['SEASON_ID'].value_counts().sort_index())
```

## Missing values in specific columns

```python
missing = matched.isna().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print(missing)
```

```python
print(len(missing))  # how many columns have missing values
```

```python
problem_row = matched[matched['AWAY_rolling_pts_5'].isna() & matched['HOME_rest_days'].notna()]
print(problem_row[['GAME_ID', 'GAME_DATE', 'HOME_TEAM_ABBR', 'AWAY_TEAM_ABBR']])
```

## Every rolling and matchup column for pts and reb

```python
print([c for c in matched.columns if 'PTS' in c.upper() or 'REB' in c.upper()])
```
