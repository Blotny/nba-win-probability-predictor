```python
import pandas as pd
matched = pd.read_csv('../data/processed/games_final.csv')
matched['GAME_DATE'] = pd.to_datetime(matched['GAME_DATE'])
```

```python
print("Duplicates GAME_ID:", matched['GAME_ID'].duplicated().sum())
print("Date range:", matched['GAME_DATE'].min(), "-", matched['GAME_DATE'].max())
print("Number of unique seasons:", matched['SEASON_ID'].nunique())
print(matched['SEASON_ID'].value_counts().sort_index())
```

```python
missing = matched.isna().sum()
missing = missing[missing > 0].sort_values(ascending=False)
print(missing)
```
